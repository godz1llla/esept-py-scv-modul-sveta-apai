## 1-бөлім. Деректерді бастапқы зерттеу (1–5 тапсырмалар)

### 1. Өзіңіздің CSV файлыңызды жүктеп, алғашқы 10 жолын экранға шығарыңыз.
**Әдістер:** `read_csv(), head()`

**Түсіндірме:** Бұл тапсырмада екі негізгі әдіс қолданылды:
• read_csv('students_data.csv') — университет студенттерінің үлгерімі жазылған CSV файлды оқып, Pandas кітапханасының DataFrame (кестелік) құрылымына жүктейді.
• head(10) — жүктелген кестенің ең алғашқы 10 жолын (0-ден 9-ға дейінгі индекстерді) экранға шығарады.
CSV бойынша талдау: Кестеде 12 баған бар екені көрінеді: Студент_ID, Аты_жөні, Факультет, Мамандық, Курс, Қала, Оқу_түрі (Грант/Ақылы), Қатысу_пайызы, Аралық_бақылау, Емтихан_балы, GPA және Стипендия.

```python
import pandas as pd

myData = pd.read_csv("students_data.csv")
print(myData.head(10))
```

### 2. Жолдар мен бағандардың санын анықтап, баған атауларын шығарыңыз.
**Әдістер:** `shape, columns`

**Түсіндірме:** Бұл тапсырмада кестенің құрылымын анықтайтын екі атрибут қолданылды:
• shape — DataFrame өлшемін (жолдар_саны, бағандар_саны) кортежі ретінде қайтарады. Біздің студенттер деректер жиыны 600 жолдан (студенттен) және 12 бағаннан тұрады.
• columns — кестедегі барлық 12 бағанның атауларын қайтарады, ал tolist() әдісі оларды Python тізімі ретінде шығарады.

```python
print("Кесте өлшемі (shape):", myData.shape)
print("Жолдар саны:", myData.shape[0])
print("Бағандар саны:", myData.shape[1])
print("Баған атаулары:", myData.columns.tolist())
```

### 3. Деректер типтерін және әр бағандағы бос емес мәндердің санын анықтаңыз.
**Әдістер:** `info(), count()`

**Түсіндірме:** • info() әдісі кесте туралы толық техникалық мәлімет береді: жазбалар саны (600), бағандар тізімі, бос емес мәндер саны (Non-Null Count) және деректер типі (Dtype).
• count() әдісі әр бағандағы бос емес (NaN емес) мәндер санын жеке есептейді.
CSV бойынша талдау: Барлық 12 бағанда да 600 бос емес мән бар. 6 баған (Студент_ID, Курс, Қатысу_пайызы, Аралық_бақылау, Емтихан_балы, Стипендия) бүтін сандық int64 типінде, 1 баған (GPA) нақты сандық float64 типінде, ал 5 баған (Аты_жөні, Факультет, Мамандық, Қала, Оқу_түрі) мәтіндік (str) типінде сақталған.

```python
myData.info()
print("\nӘр бағандағы бос емес мәндер саны:")
print(myData.count())
```

### 4. Сандық бағандардың негізгі статистикалық сипаттамаларын алыңыз.
**Әдістер:** `describe()`

**Түсіндірме:** • describe() әдісі кестедегі барлық сандық бағандар бойынша негізгі статистикалық көрсеткіштерді (саны — count, орташа мән — mean, стандартты ауытқу — std, минимум — min, 25%, 50% медиана, 75% квартильдер және максимум — max) автоматты түрде есептейді.
CSV бойынша талдау: Студенттердің орташа сабаққа қатысуы — 83.59% (58%-дан 100%-ға дейін), аралық бақылаудың орташа балы — 77.72, емтиханның орташа балы — 79.45 (56-ден 100-ге дейін), ал орташа GPA — 2.98 (1.99-ден 3.96-ге дейін).

```python
print(myData.describe().round(2))
```

### 5. Әр бағандағы бірегей мәндердің санын анықтаңыз.
**Әдістер:** `nunique(), unique()`

**Түсіндірме:** • nunique() әдісі әр бағанда қанша қайталанбайтын (бірегей) мән бар екенін санайды.
• unique() әдісі таңдалған санаттық бағанның нақты қандай бірегей мәндерден тұратынын массив ретінде шығарады.
CSV бойынша талдау: Университет деректерінде 5 факультет, 8 мамандық, 4 курс (1–4), 6 қала, 2 оқу түрі (Грант, Ақылы) және 3 шәкіртақы деңгейі (0, 41898, 48183 тг) бар.

```python
print("Әр бағандағы бірегей мәндер саны:")
print(myData.nunique())
print("\nФакультеттер:", myData["Факультет"].unique())
print("Мамандықтар:", myData["Мамандық"].unique())
print("Қалалар:", myData["Қала"].unique())
print("Оқу түрі:", myData["Оқу_түрі"].unique())
```

## 2-бөлім. Деректерді тазалау және алдын ала өңдеу (6–10 тапсырмалар)

### 6. Бос мәндерді тауып, әр бағандағы олардың санын есептеңіз.
**Әдістер:** `isnull(), sum()`

**Түсіндірме:** • isnull() әдісі кестедегі әрбір ұяшықты тексеріп, егер мән бос (NaN) болса True, ал толық болса False қайтарады.
• sum() әдісі сол True (1-ге тең) мәндерін бағандар бойынша қосып, әр бағанда неше бос ұяшық бар екенін шығарады.
CSV бойынша талдау: Біздің students_data.csv файлында барлық 12 баған бойынша бос мәндер саны 0-ге тең, яғни деректер толық әрі таза.

```python
print("Әр бағандағы бос (NaN) мәндер саны:")
print(myData.isnull().sum())
print("Жалпы бос мәндер саны:", myData.isnull().sum().sum())
```

### 7. Бос мәндерді жою немесе тиісті әдіспен толтыру.
**Әдістер:** `dropna(), fillna()`

**Түсіндірме:** • dropna() — кемінде бір ұяшығы бос (NaN) болған жолдарды кестеден толықтай өшіреді.
• fillna() — бос ұяшықтарды өшірмей, оларды белгілі бір мәнмен (мысалы, сандық 'GPA' бағанын орташа мәнмен, ал мәтіндік 'Факультет' бағанын 'Белгісіз' сөзімен) толтырады.
CSV бойынша талдау: Бастапқы деректерде бос ұяшықтар болмағандықтан, dropna() әдісінен кейін де барлық 600 жол мен 12 баған өзгеріссіз сақталды.

```python
clean_drop = myData.dropna()
clean_fill = myData.copy()
clean_fill["GPA"] = clean_fill["GPA"].fillna(clean_fill["GPA"].mean())
clean_fill["Факультет"] = clean_fill["Факультет"].fillna("Белгісіз")
print("dropna() кейінгі кесте өлшемі:", clean_drop.shape)
print("fillna() кейінгі бос мәндер саны:", clean_fill.isnull().sum().sum())
```

### 8. Қайталанатын жолдарды анықтап, оларды жойыңыз.
**Әдістер:** `duplicated(), drop_duplicates()`

**Түсіндірме:** • duplicated() — кестедегі жолдардың бұрынғы жолдармен толықтай қайталанатынын (дубликат екенін) тексереді, ал sum() олардың жалпы санын есептейді.
• drop_duplicates() — қайталанатын артық жолдарды өшіріп, тек бірегей жазбаларды қалдырады.
CSV бойынша талдау: Кестедегі әрбір студент жазбасы бірегей болғандықтан, қайталанатын жолдар саны 0 болды.

```python
dup_count = myData.duplicated().sum()
print("Қайталанатын (дубликат) жолдар саны:", dup_count)
myData_no_dup = myData.drop_duplicates()
print("Дубликаттарды жойғаннан кейінгі өлшем:", myData_no_dup.shape)
```

### 9. Бір немесе бірнеше бағанның деректер типін өзгертіңіз.
**Әдістер:** `astype(), to_numeric()`

**Түсіндірме:** • astype(float) — 'Емтихан_балы' бағанының деректер типін бүтін саннан (int64) нақты бөлшек санға (float64) түрлендіреді.
• pd.to_numeric(..., errors='coerce') — баған мәндерін сандық типке ауыстырады (егер мәтіндік қате мәндер кездессе, оларды автоматты түрде NaN етеді).
CSV бойынша талдау: 'Емтихан_балы' бағаны int64 типінен float64 типіне сәтті ауыстырылды.

```python
myData_types = myData.copy()
myData_types["Емтихан_балы"] = myData_types["Емтихан_балы"].astype(float)
myData_types["GPA"] = pd.to_numeric(myData_types["GPA"], errors="coerce")
print("Өзгертілген деректер типтері:")
print(myData_types[["Емтихан_балы", "GPA", "Курс"]].dtypes)
```

### 10. Бағандардың атауларын өзгерту және қажетсіз бағандарды жою.
**Әдістер:** `rename(), drop()`

**Түсіндірме:** • rename(columns={'Оқу_түрі': 'Оқу_негізі'}) — бағанның атауын жаңа атауға ('Оқу_негізі') өзгертеді.
• drop(columns=['Студент_ID']) — талдауға қажет емес реттік 'Студент_ID' бағанын кестеден өшіреді.
CSV бойынша талдау: 'Студент_ID' бағаны жойылғаннан кейін кестеде 12 бағанның орнына 11 мазмұндық баған қалды.

```python
myData_renamed = myData.rename(columns={"Оқу_түрі": "Оқу_негізі"})
myData_dropped = myData_renamed.drop(columns=["Студент_ID"])
print("Жаңартылған бағандар саны:", myData_dropped.shape[1])
print("Жаңартылған баған атаулары:", myData_dropped.columns.tolist())
print(myData_dropped.head(5))
```

## 3-бөлім. Деректерді сүзгілеу және сұрыптау (11–15 тапсырмалар)

### 11. Сандық көрсеткіші (мысалы, GPA) орташа мәннен жоғары жолдарды таңдаңыз.
**Әдістер:** `Шарттар, mean()`

**Түсіндірме:** • myData['GPA'].mean() — барлық 600 студенттің орташа GPA көрсеткішін есептейді (нәтижесі: 2.98).
• myData[myData['GPA'] > mean_gpa] — логикалық шарт арқылы тек GPA балы 2.98-ден жоғары студенттерді сүзіп алады.
CSV бойынша талдау: Барлығы 600 студенттің ішінен 304 студенттің (50.7%) үлгерім балы университет бойынша орташа деңгейден жоғары.

```python
mean_gpa = myData["GPA"].mean()
above_avg = myData[myData["GPA"] > mean_gpa]
print(f"Орташа GPA: {mean_gpa:.2f}")
print(f"GPA көрсеткіші орташадан жоғары студенттер саны: {len(above_avg)} (600 ішінен)")
print(above_avg.head(10))
```

### 12. Екі немесе одан да көп шартқа сәйкес келетін жазбаларды іріктеңіз.
**Әдістер:** `loc[], логикалық операторлар (&, |)`

**Түсіндірме:** • loc[] — жолдар мен бағандарды белгіленген логикалық шарттар бойынша іріктеп алу индексаторы.
• & (ЖӘНЕ) операторы барлық шарттардың бір мезгілде орындалуын талап етеді, ал | (НЕМЕСЕ) операторы кемінде бір шарттың орындалуын тексереді.
CSV бойынша талдау: Ақпараттық технологиялар факультетінде грантта оқитын әрі GPA балы 3.2-ден асатын 36 үздік студент табылды. Ал Алматы немесе Астана қалаларынан келген студенттер саны — 291.

```python
# 1) & (ЖӘНЕ) операторы: Ақпараттық технологиялар факультеті + Грант + GPA > 3.2
filter_and = myData.loc[
    (myData["Факультет"] == "Ақпараттық технологиялар") &
    (myData["Оқу_түрі"] == "Грант") &
    (myData["GPA"] > 3.2)
]
print("1) АТ факультетінде грантта оқитын және GPA > 3.2 студенттер саны:", len(filter_and))

# 2) | (НЕМЕСЕ) операторы: Алматы НЕМЕСЕ Астана қаласынан келген студенттер
filter_or = myData.loc[(myData["Қала"] == "Алматы") | (myData["Қала"] == "Астана")]
print("2) Алматы немесе Астана қаласынан келген студенттер саны:", len(filter_or))
print(filter_and.head(8))
```

### 13. Таңдалған көрсеткіш бойынша ең үлкен мәндері бар 10 жазбаны табыңыз.
**Әдістер:** `nlargest()`

**Түсіндірме:** • nlargest(10, 'Емтихан_балы') — деректерді толық сұрыптамай-ақ, 'Емтихан_балы' бағаны бойынша ең жоғары 10 жазбаны кему ретімен жылдам таңдап береді.
CSV бойынша талдау: Емтиханнан ең жоғары балл (100 балл) жинаған ТОП-10 студенттің тізімі шығарылды. Олардың GPA көрсеткіштері де өте жоғары (3.55 – 4.00 аралығында).

```python
top10_exam = myData.nlargest(10, "Емтихан_балы")
print(top10_exam[["Студент_ID", "Аты_жөні", "Факультет", "Мамандық", "Курс", "Қала", "Емтихан_балы", "GPA", "Стипендия"]])
```

### 14. Ең кіші мәндері бар 5 жазбаны табыңыз.
**Әдістер:** `nsmallest()`

**Түсіндірме:** • nsmallest(5, 'GPA') — 'GPA' бағаны бойынша ең төменгі 5 жазбаны өсу ретімен табады.
CSV бойынша талдау: Ең төменгі GPA көрсеткіші — 1.99 балл (Аяулым Мұқанова, Инженерия). Кестеден бұл студенттердің сабаққа қатысу пайызы да төмен (58%–68%) екенін байқауға болады.

```python
bottom5_gpa = myData.nsmallest(5, "GPA")
print(bottom5_gpa[["Студент_ID", "Аты_жөні", "Факультет", "Мамандық", "Курс", "Қала", "Қатысу_пайызы", "Емтихан_балы", "GPA"]])
```

### 15. Деректерді бір немесе бірнеше баған бойынша сұрыптаңыз.
**Әдістер:** `sort_values()`

**Түсіндірме:** • sort_values(by=['Факультет', 'GPA'], ascending=[True, False]) — кестені бірнеше баған бойынша сұрыптайды: алдымен 'Факультет' атауын әліпби ретімен (ascending=True: Ақпараттық технологиялар бірінші), ал әр факультет ішінде 'GPA' балын кему ретімен (ascending=False: ең үздігінен бастап) реттейді.
CSV бойынша талдау: Нәтижеде Ақпараттық технологиялар факультетінің ең үздік студенттері тізім басына шықты (ең жоғары GPA = 3.90).

```python
sorted_data = myData.sort_values(by=["Факультет", "GPA"], ascending=[True, False])
print(sorted_data[["Студент_ID", "Аты_жөні", "Факультет", "Мамандық", "Курс", "Емтихан_балы", "GPA", "Стипендия"]].head(10))
```

## 4-бөлім. Статистикалық талдау (16–20 тапсырмалар)

### 16. Сандық көрсеткіштердің орташа мәнін, медианасын, минимумын және максимумын есептеңіз.
**Әдістер:** `mean(), median(), min(), max()`

**Түсіндірме:** • mean() — арифметикалық орташа мәнді, median() — реттелген қатардың дәл ортасындағы медиананы, min() — ең кіші мәнді, max() — ең үлкен мәнді есептейді.
CSV бойынша талдау: Барлық сандық бағандарда орташа мән мен медиана бір-біріне өте жақын (мысалы, GPA орташасы — 2.98, медианасы — 2.98; Емтихан балының орташасы — 79.45, медианасы — 79.00). Бұл студенттер үлгерімінің қалыпты (симметриялы) таралғанын дәлелдейді.

```python
num_cols = ["Курс", "Қатысу_пайызы", "Аралық_бақылау", "Емтихан_балы", "GPA", "Стипендия"]
stats_16 = pd.DataFrame({
    "Орташа мән (mean)": myData[num_cols].mean().round(2),
    "Медиана (median)": myData[num_cols].median().round(2),
    "Минимум (min)": myData[num_cols].min(),
    "Максимум (max)": myData[num_cols].max()
})
print(stats_16)
```

### 17. Стандартты ауытқу мен дисперсияны есептеңіз.
**Әдістер:** `std(), var()`

**Түсіндірме:** • std() (стандартты ауытқу) және var() (дисперсия, яғни std квадраты) — деректердің орташа мәннен қаншалықты алшақ шашырағанын көрсетеді.
CSV бойынша талдау: GPA бойынша стандартты ауытқу небәрі 0.38 балды құрайды (дисперсиясы 0.14), ал Емтихан балы бойынша std = 8.44 балл.

```python
stats_17 = pd.DataFrame({
    "Стандартты ауытқу (std)": myData[num_cols].std().round(2),
    "Дисперсия (var)": myData[num_cols].var().round(2)
})
print(stats_17)
```

### 18. Таңдалған көрсеткіштің 25, 50 және 75-процентильдерін табыңыз.
**Әдістер:** `quantile()`

**Түсіндірме:** • quantile([0.25, 0.50, 0.75]) — деректерді төрт тең бөлікке бөлетін квартильдік шектерді (25%, 50% медиана, 75%) есептейді.
CSV бойынша талдау: Студенттердің 25%-ында GPA 2.73-ден төмен, 50%-ында (медиана) 2.98-ден төмен, ал ең үздік 25% студенттің GPA көрсеткіші 3.23-ден жоғары (4.00-ге дейін).

```python
q_table = myData[["GPA", "Емтихан_балы", "Қатысу_пайызы"]].quantile([0.25, 0.50, 0.75])
print(q_table)
```

### 19. Ең жиі кездесетін мәнді анықтаңыз.
**Әдістер:** `mode(), value_counts()`

**Түсіндірме:** • mode() — бағандағы ең жиі кездесетін мәнді (моданы) табады.
• value_counts() — бағандағы әрбір бірегей мәннің қанша рет кездесетінін кему ретімен санап береді.
CSV бойынша талдау: Ең көп студент оқитын факультет — 'Ақпараттық технологиялар' (209 студент), ең танымал мамандық — 'Математика' (97 студент), ең көп студент келген қала — 'Алматы' (173 студент), ал оқу түрі бойынша 'Грант' (371 студент) басым.

```python
for col in ["Факультет", "Мамандық", "Қала", "Оқу_түрі"]:
    print(f"{col} модасы: {myData[col].mode()[0]}")
    print(myData[col].value_counts())
    print("-" * 40)
```

**Түсіндірме:** • corr() — сандық бағандар арасындағы Пирсонның сызықтық корреляция коэффициенттерінің матрицасын (-1 мен +1 аралығында) есептейді.
CSV бойынша талдау: Кестеде өте күшті оң корреляция байқалады:
1) Сабаққа қатысу пайызы мен Аралық бақылау арасындағы байланыс r = 0.694, ал GPA-мен байланысы r = 0.711.

3) GPA мен Стипендия мөлшері арасында да айқын оң корреляция бар (r = 0.400), яғни сабаққа тұрақты қатысып, жоғары балл жинаған студенттер жоғары GPA мен шәкіртақыға ие болады.

```python
corr_matrix = myData[num_cols].corr().round(3)
print(corr_matrix)
```
